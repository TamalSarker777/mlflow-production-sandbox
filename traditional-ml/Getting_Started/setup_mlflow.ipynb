## Configure Tracking
MLflow supports different <b>backends</b> for tracking your experiment data. Choose one of the following options to get started. <br>

#### Option A: Database (Recommended)

Set the tracking URI to a local database URI (e.g., sqlite:///mlflow.db). This is recommended option for quickstart and <b>local development.</b>

```python
import mlflow
```

```python
mlflow.set_tracking_uri("sqlite:///mlflow.db")
mlflow.set_experiment("my_first_experiment")
```

#### Option B: File System

MLflow will automatically use local file storage if no tracking URI is specified:

```python
mlflow.set_experiment("my-first-experiment") # TO BE DEPRECATED SOON
```

#### Option C: Remote Tracking 
Start a remote MLflow tracking server following the Self Hosting Guide. Then configure your client to use the remote server:

```python
mlflow.set_tracking_uri("http://localhost:5000")
mlflow.set_experiment("my_first_experiment")


# Or we can see another way

# export MLFLOW_TRACKING_URI="http://localhost:5000"
# export MLFLOW_EXPERIMENT_NAME="my-first-experiment"
```

## Verify the connection

```python
# print the connection information
print(f"MLflow Tracking URI: ", mlflow.get_tracking_uri())
print(f"Active Experiment: ", mlflow.get_experiment_by_name('my_first_experiment'))
```

```python
with mlflow.start_run():
    mlflow.log_param("test_param", "test_value")
    print("Sucessfully connected to mlflow")
```

On Unix/Linux, multiple worker processes can cleanly inherit or share the master listening socket file descriptor using fork(). On Windows, there is no fork(); Python uses spawn, and Uvicorn's method of passing pre-bound listening sockets across multiple spawned child worker processes frequently triggers [WinError 10022] inside the Windows socket API.

```python
## if still error try:
# mlflow server --backend-store-uri sqlite:///mlflow.db --workers 1 --host 127.0.0.1 --port 5000
```

#### By setting --workers 1:
1. Uvicorn does not spawn child worker processes to divide traffic.
2. The single parent process directly creates, binds, and listens on the network socket on 127.0.0.1:5000.
3. No socket handles need to be duplicated or serialized across process boundaries, completely bypassing the Windows socket limitation.
